# The Error Function

The error function, written $\operatorname{erf}(x)$, is a smooth function defined by the integral

$$
\operatorname{erf}(x) = \frac{2}{\sqrt{\pi}}\int_0^x e^{-u^2}\,du.
$$

It is odd, with $\operatorname{erf}(0)=0$, and approaches $1$ as $x$ increases and $-1$ as $x$ decreases. The complementary error function is $\operatorname{erfc}(x)=1-\operatorname{erf}(x)$. It appears in the semi-infinite diffusion solution, where it describes how a concentration or potential changes with distance from a maintained surface.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import erf
```

```python
sample_points = [-2, -1, 0, 0.5, 1, 2]

print(f"{'x':>6} {'erf(x)':>12}")
for value in sample_points:
    print(f"{value:6.1f} {erf(value):12.6f}")
```

```python
x = np.linspace(-3, 3, 500)
erf_values = np.vectorize(erf)(x)

fig, ax = plt.subplots()
ax.plot(x, erf_values, label="erf(x)")
ax.axhline(1, color="black", linewidth=0.8, linestyle=":")
ax.axhline(-1, color="black", linewidth=0.8, linestyle=":")
ax.axvline(0, color="black", linewidth=0.8, linestyle=":")
ax.set(
    xlabel="x",
    ylabel="erf(x)",
    title="The error function",
    ylim=(-1.1, 1.1),
)
ax.grid(True, alpha=0.3)
plt.show()
```
