# Case 3: Edges with Zero Potential

Assume:
- The medium is homogeneous over a finite length $L$.
- The initial potential is uniform, $c(x,0)=c_0$.
- Potential can leave freely from both boundaries, so $c(0,t)=c(L,t)=0$ for $t>0$.

The governing equation and conditions are

$$
\frac{\partial c}{\partial t} = D\frac{\partial^2 c}{\partial x^2}, \qquad 0<x<L,\ t>0,
$$

$$
c(x,0)=c_0, \qquad c(0,t)=c(L,t)=0.
$$

The analytical solution is

$$
\frac{c(x,t)}{c_0} = \frac{4}{\pi}\sum_{m=0}^{\infty}\frac{1}{2m+1}\sin\left(\frac{(2m+1)\pi x}{L}\right)\exp\left(-\frac{(2m+1)^2\pi^2Dt}{L^2}\right).
$$

The plot shows the potential profile over the interval at several times. The legend reports $Dt$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def zero_potential_edges(x, t, D, L, c0=1.0, n_terms=100):
    """Diffusion on 0 <= x <= L with zero-potential boundaries."""
    x = np.atleast_1d(np.asarray(x, dtype=float))
    odd_modes = 2 * np.arange(n_terms) + 1
    spatial_modes = np.sin(np.outer(odd_modes * np.pi / L, x))
    decay = np.exp(-(odd_modes**2) * np.pi**2 * D * t / L**2)
    amplitudes = 4 * c0 / (np.pi * odd_modes)
    return np.sum(amplitudes[:, None] * spatial_modes * decay[:, None], axis=0)
```

```python
D = 1.0
L = 2.0
c0 = 1.0
times = [0.1, 0.5, 1.0, 2.0]

x = np.linspace(0, L, 500)

fig, ax = plt.subplots()
for t in times:
    ax.plot(
        x,
        zero_potential_edges(x, t, D, L, c0),
        label=f"Dt = {D * t:g}",
    )

ax.axhline(c0, color="black", linewidth=0.8, linestyle=":")
ax.set(
    xlabel="Position x",
    ylabel="Potential c(x, t)",
    title="Diffusion with zero-potential edges",
    xlim=(0, L),
    ylim=(0, 1.05),
)
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
