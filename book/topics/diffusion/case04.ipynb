# Case 4: Radial Coordinates
The diffusion equation can be written in radial coordinates as

$$
\frac{\partial c}{\partial t} = D\left(\frac{\partial^2 c}{\partial r^2} + \frac{1}{r}\frac{\partial c}{\partial r} + \frac{1}{r^2}\frac{\partial^2 c}{\partial \theta^2}\right), \qquad 0<r<\infty,\ 0\le\theta<2\pi,\ t>0.
$$

Assume:

- Radial coordinates $r$ and $\theta$,over which the medium is homogeneous.
- The initial potential is uniform, $c(r,\theta,0)=c_0$.
- Potential can leave freely at the center of the domain with constant flow $Q$ for $t>0$ (i.e., a source or sink in storage per unit time).
- Potential $c(\infty,\theta,t)=c_0$ for $t>0$.

The 1D analytic solution in the radial direction is:

$$
c(r,t) = c_0 + \frac{Q}{4\pi D c_0} \int_0^t \frac{1}{t'} \exp\left(-\frac{r^2}{4Dt'}\right) \, dt'
$$

The plot shows the potential profile over the interval at several times. The legend reports $Dt$.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def zero_potential_edges(x, t, D, L, c0=1.0, n_terms=100):
    """Diffusion on 0 <= x <= L with zero-potential boundaries."""
    x = np.atleast_1d(np.asarray(x, dtype=float))
    odd_modes = 2 * np.arange(n_terms) + 1
    spatial_modes = np.sin(np.outer(odd_modes * np.pi / L, x))
    decay = np.exp(-(odd_modes**2) * np.pi**2 * D * t / L**2)
    amplitudes = 4 * c0 / (np.pi * odd_modes)
    return np.sum(amplitudes[:, None] * spatial_modes * decay[:, None], axis=0)
```

```python
D = 1.0
L = 2.0
c0 = 1.0
times = [0.1, 0.5, 1.0, 2.0]

x = np.linspace(0, L, 500)

fig, ax = plt.subplots()
for t in times:
    ax.plot(
        x,
        zero_potential_edges(x, t, D, L, c0),
        label=f"Dt = {D * t:g}",
    )

ax.axhline(c0, color="black", linewidth=0.8, linestyle=":")
ax.set(
    xlabel="Position x",
    ylabel="Potential c(x, t)",
    title="Diffusion with zero-potential edges",
    xlim=(0, L),
    ylim=(0, 1.05),
)
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
