# Case 2: Plane Source

For a plane source in 1D, the essential boundary conditions and assumptions are:
- The medium is homogeneous and infinite in extent.
- The initial potential is zero everywhere except at the plane source.
- The plane source releases a storage quantity $S_0$ instantaneously at $x=0$ and $t=0$.

The solution, found by inspection, is:

$$
c(x,t) = \frac{S_0}{\sqrt{4 \pi D t}} \exp\left(-\frac{x^2}{4 D t}\right)
$$

$S$ is the storage quantity of the diffusing substance.

For plane source with $+x$ direction only, reflect and superimpose, the solution is:

$$
c(x,t) = \frac{S_0}{\sqrt{\pi D t}} \exp\left(-\frac{x^2}{4 D t}\right)
$$

Plots are for $Dt$.



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plane_source_1d(x, t, D, S0):
    """Plane source in an infinite 1D medium."""
    return S0 / np.sqrt(4 * np.pi * D * t) * np.exp(-(x**2) / (4 * D * t))


def plane_source_positive_x(x, t, D, S0):
    """Plane source in an infinite 1D medium with reflection at x=0 for the positive-x case."""
    return S0 / np.sqrt(np.pi * D * t) * np.exp(-(x**2) / (4 * D * t))
```

```python
D = 1.0
S0 = 1.0
times = [0.1, 0.5, 1.0, 2.0]

# Finite plotting window approximating the infinite spatial domain.
x = np.linspace(-10, 10, 1000)

fig, ax = plt.subplots()
for t in times:
    ax.plot(x, plane_source_1d(x, t, D, S0), label=f"Dt = {D * t:g}")

ax.set(
    xlabel="Position x",
    ylabel="Concentration c(x, t)",
    title="Plane source in an infinite 1D medium",
)
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

Now we will look at the second case

```python
x_positive = np.linspace(0, 10, 500)

fig, ax = plt.subplots()
for t in times:
    ax.plot(
        x_positive,
        plane_source_positive_x(x_positive, t, D, S0),
        label=f"Dt = {D * t:g}",
    )

ax.set(
    xlabel="Position x",
    ylabel="Concentration c(x, t)",
    title="Plane source in the positive-x direction",
)
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

Now both plotted together.

```python
D = 1.0
t = 1.0

fig, ax = plt.subplots()
ax.plot(x_positive, plane_source_1d(x_positive, t, D, S0), label="Infinite 1D")
ax.plot(x_positive, plane_source_positive_x(x_positive, t, D, S0), label="Positive-x")

ax.set(
    xlabel="Position x",
    ylabel="Concentration c(x, t)",
    title=f"Both plane-source methods (D={D}, t={t})",
)
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
